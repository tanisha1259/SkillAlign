```python
# Count all JSON examples by filename category

from collections import Counter

categories = Counter()

for f in json_files:
    name = f.name.lower()

    if name.startswith("match_"):
        categories["match"] += 1
    elif "mismatch" in name:
        categories["mismatch"] += 1
    elif "invalid" in name:
        categories["invalid"] += 1
    elif "gibberish" in name:
        categories["gibberish"] += 1
    elif "empty_additional_info" in name:
        categories["empty_additional_info"] += 1
    else:
        categories["other"] += 1

print("Dataset categories:")
for category, count in categories.items():
    print(f"{category}: {count}")

print("\nTotal:", sum(categories.values()))
```

```python
import json
from collections import Counter

true_mismatch_files = [
    f for f in json_files
    if f.name.startswith("mismatch_")
]

valid_count = 0
invalid_count = 0

requirement_counts = Counter()
meets_counts = Counter()

for f in true_mismatch_files:
    with open(f, "r", encoding="utf-8") as file:
        data = json.load(file)

    valid = data.get("output", {}).get("valid_resume_and_jd", False)

    if valid:
        valid_count += 1
    else:
        invalid_count += 1

    requirements = (
        data.get("output", {})
            .get("scores", {})
            .get("requirements", [])
    )

    for req in requirements:
        criterion = req.get("criteria")
        meets = req.get("meets")

        requirement_counts[criterion] += 1
        meets_counts[str(meets)] += 1

print("Total mismatch files:", len(true_mismatch_files))
print("Valid:", valid_count)
print("Invalid:", invalid_count)

print("\nRequirement labels:")
for k, v in meets_counts.items():
    print(k, ":", v)

print("\nTotal requirement annotations:", sum(requirement_counts.values()))

print("\nMost common requirements:")
for req, count in requirement_counts.most_common(20):
    print(count, "->", req)
```

```python
import json
from collections import Counter, defaultdict

valid_match_files = []
valid_mismatch_files = []

for f in json_files:

    if f.name.startswith("match_") or f.name.startswith("mismatch_"):

        with open(f, "r", encoding="utf-8") as file:
            data = json.load(file)

        if data.get("output", {}).get("valid_resume_and_jd") is True:

            if f.name.startswith("match_"):
                valid_match_files.append(f)
            else:
                valid_mismatch_files.append(f)


print("Valid match files:", len(valid_match_files))
print("Valid mismatch files:", len(valid_mismatch_files))


# --------------------------------------------------
# Requirement-level statistics
# --------------------------------------------------

label_counts = Counter()
requirement_counts = Counter()

# Track which requirements have both labels
requirement_labels = defaultdict(set)

total_requirements = 0

all_valid_files = valid_match_files + valid_mismatch_files

for f in all_valid_files:

    with open(f, "r", encoding="utf-8") as file:
        data = json.load(file)

    requirements = (
        data.get("output", {})
            .get("scores", {})
            .get("requirements", [])
    )

    for req in requirements:

        criterion = req.get("criteria")
        meets = req.get("meets")

        if criterion is None or meets is None:
            continue

        total_requirements += 1

        label_counts[str(meets)] += 1
        requirement_counts[criterion] += 1
        requirement_labels[criterion].add(meets)


print("\nTotal requirement annotations:", total_requirements)

print("\nLabel distribution:")
for label, count in label_counts.items():
    print(f"{label}: {count}")


# --------------------------------------------------
# Requirements appearing with BOTH True and False
# --------------------------------------------------

both_labels = [
    req
    for req, labels in requirement_labels.items()
    if True in labels and False in labels
]

only_true = [
    req
    for req, labels in requirement_labels.items()
    if labels == {True}
]

only_false = [
    req
    for req, labels in requirement_labels.items()
    if labels == {False}
]


print("\nUnique requirements:", len(requirement_counts))

print("Requirements with BOTH labels:", len(both_labels))
print("Requirements with ONLY True:", len(only_true))
print("Requirements with ONLY False:", len(only_false))


print("\nExamples appearing with BOTH labels:")

for req in sorted(both_labels, key=lambda x: -requirement_counts[x])[:30]:
    print(
        f"{requirement_counts[req]:3} | "
        f"{req}"
    )
```

```python
# --------------------------------------------------
# Inspect examples for common requirements
# --------------------------------------------------

target_requirements = [
    "5+ years of experience in business development or sales",
    "Strong understanding of IT and manufacturing industries",
    "Strong proficiency in Microsoft Office Suite",
    "Bachelor's degree in Business Administration or related field"
]


for target in target_requirements:

    print("\n" + "=" * 80)
    print("REQUIREMENT:", target)
    print("=" * 80)

    found = 0

    for f in all_valid_files:

        with open(f, "r", encoding="utf-8") as file:
            data = json.load(file)

        requirements = (
            data.get("output", {})
                .get("scores", {})
                .get("requirements", [])
        )

        for req in requirements:

            if req.get("criteria") == target:

                print("\nFile:", f.name)
                print("Meets:", req.get("meets"))

                found += 1

                if found >= 5:
                    break

        if found >= 5:
            break
```

```python
import json
from collections import Counter

resume_ids = []
job_ids = []

for f in all_valid_files:

    with open(f, "r", encoding="utf-8") as file:
        data = json.load(file)

    # NETSOL may not explicitly expose IDs,
    # so use filename + personal information where available
    personal = data.get("output", {}).get("personal_info", {})

    name = personal.get("name")
    email = personal.get("email")

    job_description = (
        data.get("input", {})
            .get("job_description", "")
    )

    resume = (
        data.get("input", {})
            .get("resume", "")
    )

    resume_ids.append(
        email if email else name if name else resume[:100]
    )

    job_ids.append(job_description)


resume_counts = Counter(resume_ids)
job_counts = Counter(job_ids)

print("Unique candidate identities:", len(resume_counts))
print("Total examples:", len(resume_ids))

print(
    "Candidates appearing more than once:",
    sum(1 for x in resume_counts.values() if x > 1)
)

print(
    "Maximum examples for one candidate:",
    max(resume_counts.values())
)

print("\nJob descriptions:")
print("Unique JDs:", len(job_counts))
print(
    "JDs appearing more than once:",
    sum(1 for x in job_counts.values() if x > 1)
)

print(
    "Maximum examples for one JD:",
    max(job_counts.values())
)
```

## 4. NETSOL Dataset — Clean Requirement-Level Dataset

```python
from huggingface_hub import snapshot_download
from pathlib import Path
import json
import pandas as pd

repo_path = snapshot_download(
    repo_id="netsol/resume-score-details",
    repo_type="dataset"
)

json_files = list(Path(repo_path).rglob("*.json"))

print("Dataset path:")
print(repo_path)

print("\nTotal JSON files:", len(json_files))
```

```python
valid_files = []

for f in json_files:

    if not (
        f.name.startswith("match_")
        or f.name.startswith("mismatch_")
    ):
        continue

    with open(f, "r", encoding="utf-8") as file:
        data = json.load(file)

    if data.get("output", {}).get("valid_resume_and_jd") is True:
        valid_files.append(f)

print("Valid NETSOL files:", len(valid_files))
```

```python
rows = []

for f in valid_files:

    with open(f, "r", encoding="utf-8") as file:
        data = json.load(file)

    input_data = data.get("input", {})
    output_data = data.get("output", {})

    resume = input_data.get("resume", "")
    job_description = input_data.get("job_description", "")

    personal_info = output_data.get("personal_info", {})

    # Temporary identity for leakage analysis.
    # We will NOT use this as a model feature.
    candidate_id = personal_info.get("email")

    requirements = (
        output_data
        .get("scores", {})
        .get("requirements", [])
    )

    for req in requirements:

        criterion = req.get("criteria")
        meets = req.get("meets")

        if criterion is None or meets is None:
            continue

        rows.append({
            "source_file": f.name,
            "candidate_id": candidate_id,
            "resume": resume,
            "job_description": job_description,
            "requirement": criterion,
            "label": int(meets)
        })

netsol_df = pd.DataFrame(rows)

print("Total requirement annotations:", len(netsol_df))
print("\nColumns:")
print(netsol_df.columns.tolist())
```

```python
print("Label distribution:")
print(netsol_df["label"].value_counts())

print("\nUnique candidates:")
print(netsol_df["candidate_id"].nunique())

print("\nUnique requirements:")
print(netsol_df["requirement"].nunique())

print("\nMissing values:")
print(netsol_df.isna().sum())
```

```python
duplicate_mask = netsol_df.duplicated(
    subset=[
        "resume",
        "job_description",
        "requirement"
    ],
    keep=False
)

duplicates = netsol_df[duplicate_mask]

print("Duplicate rows:", len(duplicates))
```

```python
conflicts = (
    netsol_df
    .groupby(
        ["resume", "job_description", "requirement"]
    )["label"]
    .nunique()
)

conflicts = conflicts[conflicts > 1]

print(
    "Conflicting resume/JD/requirement examples:",
    len(conflicts)
)
```

```python
conflict_keys = conflicts.index

conflict_examples = netsol_df.merge(
    conflict_keys.to_frame(index=False),
    on=["resume", "job_description", "requirement"],
    how="inner"
)

print("Conflict rows:", len(conflict_examples))

print(
    conflict_examples[
        [
            "source_file",
            "candidate_id",
            "requirement",
            "label"
        ]
    ].sort_values(
        ["requirement", "candidate_id", "label"]
    ).to_string(index=False)
)
```

```python
conflict_summary = (
    conflict_examples
    .groupby(
        ["candidate_id", "requirement"]
    )["label"]
    .agg(
        labels=lambda x: sorted(set(x)),
        count="count"
    )
    .reset_index()
)

print(conflict_summary.to_string(index=False))
```

```python
for i, (_, row) in enumerate(
    conflict_examples.drop_duplicates(
        subset=["resume", "job_description", "requirement"]
    ).head(5).iterrows()
):

    print("\n" + "=" * 100)
    print("CONFLICT", i + 1)
    print("=" * 100)

    print("\nRequirement:")
    print(row["requirement"])

    print("\nResume:")
    print(row["resume"][:3000])

    print("\nJob Description:")
    print(row["job_description"][:3000])

    print("\nFiles containing this example:")

    matching = netsol_df[
        (netsol_df["resume"] == row["resume"]) &
        (netsol_df["job_description"] == row["job_description"]) &
        (netsol_df["requirement"] == row["requirement"])
    ]

    print(
        matching[
            ["source_file", "candidate_id", "label"]
        ].to_string(index=False)
    )
```

```python
conflict_keys = (
    netsol_df
    .groupby(
        ["resume", "job_description", "requirement"]
    )["label"]
    .nunique()
)

conflict_keys = conflict_keys[
    conflict_keys > 1
]

print("Conflicting groups:", len(conflict_keys))
```

```python
clean_df = netsol_df.merge(
    conflict_keys.reset_index()[
        ["resume", "job_description", "requirement"]
    ],
    on=["resume", "job_description", "requirement"],
    how="left",
    indicator=True
)

clean_df = clean_df[
    clean_df["_merge"] == "left_only"
].drop(columns="_merge")

print("Rows after removing conflicts:", len(clean_df))
```

```python
before_dedup = len(clean_df)

clean_df = clean_df.drop_duplicates(
    subset=[
        "resume",
        "job_description",
        "requirement",
        "label"
    ]
).reset_index(drop=True)

after_dedup = len(clean_df)

print("Before deduplication:", before_dedup)
print("After deduplication:", after_dedup)
print("Duplicate rows removed:", before_dedup - after_dedup)
```

```python
remaining_conflicts = (
    clean_df
    .groupby(
        ["resume", "job_description", "requirement"]
    )["label"]
    .nunique()
)

remaining_conflicts = remaining_conflicts[
    remaining_conflicts > 1
]

print(
    "Remaining conflicting groups:",
    len(remaining_conflicts)
)
```

```python
print("Final dataset size:", len(clean_df))

print("\nLabels:")
print(clean_df["label"].value_counts())

print("\nLabel percentages:")
print(
    clean_df["label"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
print("Unique candidates:", clean_df["candidate_id"].nunique())
print("Unique JDs:", clean_df["job_description"].nunique())

candidate_counts = clean_df["candidate_id"].value_counts()
jd_counts = clean_df["job_description"].value_counts()

print("\nCandidate annotations:")
print(candidate_counts.describe())

print("\nJD annotations:")
print(jd_counts.describe())

print("\nCandidates appearing in multiple examples:",
      (candidate_counts > 1).sum())

print("JDs appearing in multiple examples:",
      (jd_counts > 1).sum())
```

```python
print("\nCandidate overlap:")
print(candidate_counts.sort_values(ascending=False).head(10))

print("\nJD overlap:")
print(jd_counts.sort_values(ascending=False).head(10))
```

```python
from sklearn.model_selection import GroupShuffleSplit

# Split by candidate so the same candidate never appears
# in more than one split.

gss_train = GroupShuffleSplit(
    n_splits=1,
    test_size=0.30,
    random_state=42
)

train_idx, temp_idx = next(
    gss_train.split(
        clean_df,
        groups=clean_df["candidate_id"]
    )
)

train_df = clean_df.iloc[train_idx].reset_index(drop=True)
temp_df = clean_df.iloc[temp_idx].reset_index(drop=True)

# Split the remaining 30% equally:
# 15% validation, 15% test

gss_val_test = GroupShuffleSplit(
    n_splits=1,
    test_size=0.50,
    random_state=42
)

val_idx, test_idx = next(
    gss_val_test.split(
        temp_df,
        groups=temp_df["candidate_id"]
    )
)

val_df = temp_df.iloc[val_idx].reset_index(drop=True)
test_df = temp_df.iloc[test_idx].reset_index(drop=True)

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))
```

```python
train_candidates = set(train_df["candidate_id"])
val_candidates = set(val_df["candidate_id"])
test_candidates = set(test_df["candidate_id"])

print("\nCandidate overlap:")
print("Train ∩ Val:", len(train_candidates & val_candidates))
print("Train ∩ Test:", len(train_candidates & test_candidates))
print("Val ∩ Test:", len(val_candidates & test_candidates))
```

```python
print("\nTrain labels:")
print(train_df["label"].value_counts(normalize=True).mul(100).round(2))

print("\nValidation labels:")
print(val_df["label"].value_counts(normalize=True).mul(100).round(2))

print("\nTest labels:")
print(test_df["label"].value_counts(normalize=True).mul(100).round(2))
```
